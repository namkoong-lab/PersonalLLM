```python
from datasets import load_dataset
import numpy as np
from tqdm import tqdm
```

```python
n_responses = 8
n_personas = 500
```

```python
best_choices = []
worst_choices = []

for persona_idx in tqdm(range(n_personas)):
    try:
        dataset = load_dataset("andrewsiah/PersonaPromptPersonalLLM_{}".format(persona_idx))["train"]
    except Exception as e:
        print(e)
        continue

    response_scores = []
    for r_idx in range(n_responses):

        scores = dataset["personaid_{}_response_{}_llama3_sfairx".format(persona_idx, r_idx+1)]
        response_scores.append(scores)
    
    response_scores = np.array(response_scores).T

    person_best = np.argmax(response_scores, -1)
    person_worst = np.argmin(response_scores, -1)

    best_choices.append(person_best)
    worst_choices.append(person_worst)
```

```python
best_choices = np.array(best_choices)
worst_choices = np.array(worst_choices)
best_choices.shape, worst_choices.shape
```

```python
np.save("../data/persona_prompt_best.npy", best_choices)
np.save("../data/persona_prompt_worst.npy", worst_choices)
```
